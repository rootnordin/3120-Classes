```python
import random

class Agent:
    def __init__(self, name= "James"):
        self.name = name
        self.inventory = []  # List of borrowed gadgets
        self.exposure_level = 0 # Higher means higher risk of detection

def borrow_gadget(self, gadget):
    """ Agent borrows a high-tech gadget."""
    self.inventory.append(gadget)

def return_gadget(self, gadget):
    """ Agent returns a gadget."""
    if gadget in self.inventory:
        self.inventory.remove(gadget)

class Gadget:
    def __init__(self, name, function):
        self.name = name
        self.function = function # E.g. "Invisibility Cloak", "Hacking device"
        self.available = True  # True if available for borrowing

class Agency:
    def __init__(self):
        self.agent = Agent("James") # Only if James is the acive agent
        self.gatget = {
            "Hacking Device": Gadget("Hacking Device", "Bypass security systems"),
            "Invisibility Cloak": Gadget("Invisibility Cloak", "Temporary invisibility for stealth missions"),
        }

    def borrow_gadget(self, gadget_name):
        """Agent James borrows a gadget if available."""
        if gadget_name in self.gadgets:
            gadget = self.gadgets[gadget_name]
            if gadget.available:
                gadget.available = False
                self.agent.borrow_gadget(gadget)
                print(f"Agent {self.agent.name} checked out '{gadget_name}' ({gadget.function}).")
            else:
                print(f"Gadget '{gadget_name}' is currently in use.")
        else:
            print("Invalid gadget.")

    def return_gadget(self, gadget_name):
        """ Handles gadget returns and calculates eposure risk."""
        if gadget_name in self.gadgets:
            gadget = self.gadgets[gadget_name]
            if gadget in self.agent.inventory:
                self.agent.return_gadget(gadget)
                gadget.available = True
                print(f"Agent {self.agent.name} returned '{gadget_name}'.")
                self.check_exposure()
            else:
                print(f"Agent {self.agent.name} does not have '{gadget_name}'.")

    def check_exposure(self):
        """Increase exposure risk if a gadget is returned late."""
        self.agent.exposure_level += 1 # Exposure risk increases on return
        consequence =random.choice([
            "Your identity has been flagged in the system.",
            "A security camera captured you near the drop point.",
            "A tracking device was founs in your returned gadget.",
            "Enemy agents have been spotted following you.",
            "Your clearance level has been temporarily revoked."
        ])
        print(f" Exposure Alert: {consequence} (exposure level: {self.agent.exposure_level})")

def main():
    agency = Agency()

    while True:
        print("\nSecret Spy Agency HQ -Agent James' Mission")
        print("1. Borrow Hacking Device")
        print("2. Borrow Invisibility Cloak")
        print("3. Return Hacking Device")
        print("4. Return Invisibility Cloak")
        print("5. View Status")
        print("6. Exit")

        choice = input("Choose an option: ")

        if choice == "1":
            agency.borrow_gadget("Hacking Device")

        elif choice == "2":
            agency.borrow_gadget("Invisibility Cloak")

        elif choice == "3":
            agency.borrow_gadget("Hacking Device")

        elif choice == "4":
            agency.return_gadget("Invisibility Cloak")

        elif choice == "5":
            agent = agency.agent
            print(f"\nAgent: {agent.name}")
            print(f"Exposure Level: {agent.exposure_level}")
            print(f"Borrowed Gadgets: {[gadget.name for gadget in agent.inventory]}")
            print("\nAvailable Gadgets:")
            for gadget in agency.gadgets.values():
                status = "Available" if gadget.available else "Checked Out"
                print(f"{gadget.name} - {gadget.function} ({status})")

        elif choice == "6":
            print("Mission Complete. Exiting...")
            break

        else:
            print("Invalid choice. Try again.")

if __name__ == "__main__":
    main()
    
    
    
    
```
